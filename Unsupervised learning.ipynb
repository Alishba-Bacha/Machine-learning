###**Data Importing And Preprocessing**

```python
import pandas as pd
from sklearn.preprocessing import StandardScaler
```

#####**Features of dataset**

```python
data = pd.read_csv('breast-cancer.csv')
data.head()
```

#####**Check Null values**

```python
print(data.isnull().sum())
```

**No Null values to handle**

######**Data description**

```python
data.info()
```

```python
data.describe()
```

#####**Scale Features**

```python
if 'diagnosis' in data.columns:
    data = data.drop('diagnosis', axis=1)

# Drop any non-numeric or irrelevant columns
if 'id' in data.columns:
    data = data.drop('id', axis=1)

# Standardize the dataset
scaler = StandardScaler()
data_scaled = scaler.fit_transform(data)

# Convert the scaled data back to a DataFrame
data_scaled = pd.DataFrame(data_scaled, columns=data.columns)
data_scaled.head()
```

####Visualize distribution of key features

**Histogram**

```python
import matplotlib.pyplot as plt
import seaborn as sns

data_scaled.hist(bins=20, figsize=(15, 10), color='blue', edgecolor='black')
plt.suptitle("Histograms of Scaled Features", fontsize=16)
plt.show()
```

**Box Plot**

```python
plt.figure(figsize=(15, 8))
sns.boxplot(data=data_scaled, orient="h", palette="Set2")
plt.title("Box Plots of Scaled Features")
plt.show()
```

**Pairwise Distribution for few features**

```python
selected_features = data_scaled.iloc[:, :5]
sns.pairplot(selected_features, diag_kind="kde", plot_kws={'alpha': 0.6})
plt.suptitle("Scatter Plots of Selected Features", y=1.02)
plt.show()
```

**Correlation Heatmap**

###**PCA**

#####**PCA for dimensionality reduction**

```python
from sklearn.decomposition import PCA
import numpy as np

# Apply PCA to reduce to 3 principal components
pca = PCA(n_components=3)
pca_result = pca.fit_transform(data_scaled)

# Create a DataFrame for the PCA results
pca_df = pd.DataFrame(pca_result, columns=['PC1', 'PC2', 'PC3'])
print("Explained Variance Ratio:", pca.explained_variance_ratio_)
print(pca_df.head())
```

####**Visualize Covariance ratio**

```python
plt.figure(figsize=(8, 5))
explained_variance = np.cumsum(pca.explained_variance_ratio_)
plt.bar(range(1, 4), pca.explained_variance_ratio_, alpha=0.7, align='center', label='Individual Explained Variance')
plt.step(range(1, 4), explained_variance, where='mid', label='Cumulative Explained Variance', color='red')
plt.xlabel('Principal Components')
plt.ylabel('Explained Variance Ratio')
plt.title('Explained Variance by PCA Components')
plt.legend()
plt.show()
```

####**2D Scatter Plot**

```python
plt.figure(figsize=(10, 6))
sns.scatterplot(x=pca_df['PC1'], y=pca_df['PC2'], alpha=0.7, s=50)
plt.title('2D PCA Scatter Plot')
plt.xlabel('Principal Component 1')
plt.ylabel('Principal Component 2')
plt.show()
```

####**3D Scatter Plot**


```python
from mpl_toolkits.mplot3d import Axes3D

# 3D Scatter plot using the first three principal components
fig = plt.figure(figsize=(10, 8))
ax = fig.add_subplot(111, projection='3d')
scatter = ax.scatter(pca_df['PC1'], pca_df['PC2'], pca_df['PC3'], alpha=0.7, s=50)
ax.set_title('3D PCA Scatter Plot')
ax.set_xlabel('Principal Component 1')
ax.set_ylabel('Principal Component 2')
ax.set_zlabel('Principal Component 3')
plt.show()
```

###**Clustering with K-means**

######**Using Elbow method for optimal clusters**

```python
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
import matplotlib.pyplot as plt

# Determine the optimal number of clusters using the Elbow Method
inertia = []
silhouette_scores = []
K = range(2, 11)
```

```python
for k in K:
    kmeans = KMeans(n_clusters=k, random_state=42)
    kmeans.fit(data_scaled)
    inertia.append(kmeans.inertia_)
    silhouette_scores.append(silhouette_score(data_scaled, kmeans.labels_))

# Plot the Elbow Method
plt.figure(figsize=(10, 5))
plt.plot(K, inertia, 'bo-', label='Inertia')
plt.xlabel('Number of Clusters (k)')
plt.ylabel('Inertia')
plt.title('Elbow Method for Optimal k')
plt.legend()
plt.show()
```

```python
plt.figure(figsize=(10, 5))
plt.plot(K, silhouette_scores, 'ro-', label='Silhouette Score')
plt.xlabel('Number of Clusters (k)')
plt.ylabel('Silhouette Score')
plt.title('Silhouette Scores for Optimal k')
plt.legend()
plt.show()
```

#####**Apply K-Means Clustering**

```python
# Based on the Elbow and Silhouette methods, choose an optimal k
optimal_k = 3
kmeans = KMeans(n_clusters=optimal_k, random_state=42)
clusters = kmeans.fit_predict(data_scaled)

# Add the cluster labels to the PCA DataFrame
pca_df['Cluster'] = clusters
print(pca_df.head())
```

```python
# Visualize clustering results on a 2D PCA scatter plot
plt.figure(figsize=(10, 6))
sns.scatterplot(x=pca_df['PC1'], y=pca_df['PC2'], hue=pca_df['Cluster'], palette='viridis', s=50, alpha=0.8)
plt.title('K-Means Clustering Results on PCA-Reduced Data')
plt.xlabel('Principal Component 1')
plt.ylabel('Principal Component 2')
plt.legend(title='Cluster')
plt.show()
```

```python
# Examine cluster centroids in the original scaled feature space
centroids = pd.DataFrame(kmeans.cluster_centers_, columns=data_scaled.columns)
print("Cluster Centroids:\n", centroids)

# If the original labels (e.g., diagnosis) are available, compare them
if 'diagnosis' in data.columns:
    original_labels = data['diagnosis']
    comparison_df = pd.DataFrame({'Cluster': clusters, 'Diagnosis': original_labels})
    print(comparison_df.groupby(['Cluster', 'Diagnosis']).size())
```

###**Aglomerative clustering**

######**Create dendrogram**

```python
from scipy.cluster.hierarchy import dendrogram, linkage
from sklearn.cluster import AgglomerativeClustering

# Perform hierarchical clustering using 'ward' linkage
linkage_matrix = linkage(data_scaled, method='ward')

# Plot the dendrogram
plt.figure(figsize=(12, 8))
dendrogram(linkage_matrix, truncate_mode='lastp', p=20, leaf_rotation=45., leaf_font_size=12.)
plt.title('Dendrogram for Agglomerative Clustering')
plt.xlabel('Sample Index or Cluster Size')
plt.ylabel('Distance')
plt.axhline(y=50, color='r', linestyle='--')  # Adjust threshold to decide clusters
plt.show()
```

#####**Apply agglomerative clustering**

```python
optimal_clusters = 3
agg_clustering = AgglomerativeClustering(n_clusters=optimal_clusters, linkage='ward')
agg_clusters = agg_clustering.fit_predict(data_scaled)

# Add the Agglomerative Clustering labels to the PCA DataFrame
pca_df['Agg_Cluster'] = agg_clusters
print(pca_df.head())
```

#####**Visualization on PCA reduced data**

```python
plt.figure(figsize=(10, 6))
sns.scatterplot(x=pca_df['PC1'], y=pca_df['PC2'], hue=pca_df['Agg_Cluster'], palette='coolwarm', s=50, alpha=0.8)
plt.title('Agglomerative Clustering Results on PCA-Reduced Data')
plt.xlabel('Principal Component 1')
plt.ylabel('Principal Component 2')
plt.legend(title='Cluster')
plt.show()
```

#####**Compare Clusters from Agglomerative Clustering and K-Means**

```python
# Compare K-Means and Agglomerative Clustering labels
comparison_df = pd.DataFrame({
    'KMeans_Cluster': pca_df['Cluster'],
    'Agg_Cluster': pca_df['Agg_Cluster']
})

# Count occurrences of each combination
comparison_table = comparison_df.groupby(['KMeans_Cluster', 'Agg_Cluster']).size().unstack(fill_value=0)
print("Cluster Comparison:\n", comparison_table)

# Plot the comparison heatmap
plt.figure(figsize=(8, 6))
sns.heatmap(comparison_table, annot=True, fmt='d', cmap='Blues', cbar=False)
plt.title('Comparison of K-Means and Agglomerative Clustering Results')
plt.xlabel('Agglomerative Clustering')
plt.ylabel('K-Means Clustering')
plt.show()
```

###**Comparison and Interpretation**

######**Cluster comparision**

```python
from sklearn.metrics import adjusted_rand_score, silhouette_score

# Compute Adjusted Rand Index (ARI) between K-Means and Agglomerative Clustering
ari_score = adjusted_rand_score(pca_df['Cluster'], pca_df['Agg_Cluster'])
print(f"Adjusted Rand Index (ARI) between K-Means and Agglomerative Clustering: {ari_score:.4f}")

# Compute Silhouette Scores for both clustering methods
kmeans_silhouette = silhouette_score(data_scaled, pca_df['Cluster'])
agg_silhouette = silhouette_score(data_scaled, pca_df['Agg_Cluster'])

print(f"Silhouette Score for K-Means Clustering: {kmeans_silhouette:.4f}")
print(f"Silhouette Score for Agglomerative Clustering: {agg_silhouette:.4f}")
```

#####**Visual comparision**

```python
fig, axes = plt.subplots(1, 2, figsize=(15, 6), sharex=True, sharey=True)

# K-Means scatter plot
sns.scatterplot(ax=axes[0], x=pca_df['PC1'], y=pca_df['PC2'], hue=pca_df['Cluster'], palette='viridis', s=50, alpha=0.8)
axes[0].set_title('K-Means Clustering')
axes[0].set_xlabel('Principal Component 1')
axes[0].set_ylabel('Principal Component 2')

# Agglomerative Clustering scatter plot
sns.scatterplot(ax=axes[1], x=pca_df['PC1'], y=pca_df['PC2'], hue=pca_df['Agg_Cluster'], palette='coolwarm', s=50, alpha=0.8)
axes[1].set_title('Agglomerative Clustering')
axes[1].set_xlabel('Principal Component 1')
axes[1].set_ylabel('Principal Component 2')

plt.tight_layout()
plt.show()
```

#####**Interpretation of Result**

```python
centroids_df = pd.DataFrame(kmeans.cluster_centers_, columns=data_scaled.columns)
print("K-Means Cluster Centroids:\n", centroids_df)

# Analyze Agglomerative Clustering grouping
agg_cluster_distribution = pca_df['Agg_Cluster'].value_counts()
print("Agglomerative Clustering Cluster Distribution:\n", agg_cluster_distribution)

# Biological pattern analysis (if diagnosis labels are available)
if 'diagnosis' in data.columns:
    original_labels = data['diagnosis']
    cluster_comparison = pd.DataFrame({
        'KMeans_Cluster': pca_df['Cluster'],
        'Agg_Cluster': pca_df['Agg_Cluster'],
        'Diagnosis': original_labels
    })

    # Group by cluster and diagnosis
    comparison_results = cluster_comparison.groupby(['KMeans_Cluster', 'Diagnosis']).size().unstack(fill_value=0)
    print("K-Means Clustering Diagnosis Distribution:\n", comparison_results)

    comparison_results_agg = cluster_comparison.groupby(['Agg_Cluster', 'Diagnosis']).size().unstack(fill_value=0)
    print("Agglomerative Clustering Diagnosis Distribution:\n", comparison_results_agg)
```

##Strength And Weaknesses
**K-Means Clustering**:
- Strengths: Efficient for large datasets, assigns clusters based on centroids.
- Weaknesses: Sensitive to outliers and requires predefined number of clusters.

**Agglomerative Clustering**:
- Strengths: Captures hierarchical relationships, does not require a predefined number of clusters.
- Weaknesses: Computationally intensive for large datasets.

**PCA's Contribution**:
- Reduced dimensions to focus on major data variations.
- Improved clustering separation and visualization.

**Conclusion**:
- K-Means yielded a better silhouette score, suggesting tighter and more distinct clusters.
- Agglomerative clustering captured some hierarchical relationships but may overlap clusters.
